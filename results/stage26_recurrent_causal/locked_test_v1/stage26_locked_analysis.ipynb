# Stage 26 locked-test audit

This notebook reviews the frozen ten-training-seed × 100-shared-test-seed evaluation. It does not train or select a checkpoint.

```python
from pathlib import Path
import json
import csv

root = Path.cwd() / 'results/stage26_recurrent_causal/locked_test_v1'
report = json.loads((root / 'stage26_locked_statistics.json').read_text())
with (root / 'stage26_locked_seed_metrics.csv').open(newline='') as stream:
    seed_metrics = list(csv.DictReader(stream))
with (root / 'stage26_locked_statistics.csv').open(newline='') as stream:
    statistics = list(csv.DictReader(stream))
with (root / 'stage26_locked_control_statistics.csv').open(newline='') as stream:
    control_statistics = list(csv.DictReader(stream))
```

```python
# Data-quality gate and declared grain
assert report['quality']['passed']
assert report['quality']['grain']['independent_training_seeds'] == 10
assert report['quality']['grain']['shared_locked_test_seeds'] == 100
assert report['quality']['grain']['policy_episode_rows'] == 1000
report['quality']['checks']
```

```python
# Ten unique frozen policies and their paired aggregate results
assert len({row['training_base_seed'] for row in seed_metrics}) == 10
assert len({row['training_seed_index'] for row in seed_metrics}) == 10
seed_metrics
```

```python
# Crossed-bootstrap intervals, exact sign-flip tests and Holm correction
statistics
```

```python
# Rule, single-dog and reset-memory locked comparisons
assert {row['comparison'] for row in control_statistics} == {
    'RECURRENT_MINUS_RULE_BASELINE',
    'RECURRENT_MINUS_SINGLE_DOG_ONLY',
    'RECURRENT_MINUS_RESET_MEMORY_EACH_DECISION',
}
control_statistics
```
